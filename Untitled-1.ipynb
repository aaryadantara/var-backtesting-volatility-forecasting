## 2. Returns and why normality is a problem

A normal distribution VaR assumes thin tails. Equity and bonds returns usually have fat tails (high excess kurtosis), so a normal model should understate extreme losses. Thr table chekcs some basic stats on the data.

```python
returns = np.log(prices/prices.shift(1)).dropna()
returns
```

```python
desc = pd.DataFrame({
    "Daily mean (%)": returns.mean() * 100,
    "Daily std (%)": returns.std() * 100,
    "Skewness": returns.apply(stats.skew),
    "Excess kurtosis": returns.apply(stats.kurtosis),
    "Worst day (%)": returns.min() * 100,
    "Best day (%)": returns.max() * 100,
    "Jarque-Bera p": returns.apply(lambda r: stats.jarque_bera(r).pvalue)
})

display(desc.round(4))
```

## 3. Models and forecasts

Four VaR forecasts, all 1-day at 99%, each using only data available the day before:

| Model | Idea |
|---|---|
| `historical` | Empirical 1% quantile of the last 1,000 returns |
| `ewma` | RiskMetrics: $\sigma_t^2 = \lambda\,\sigma_{t-1}^2 + (1-\lambda)\,r_{t-1}^2$, with a normal quantile |
| `garch` | GARCH(1,1) volatility with a normal quantile, refit every 20 days |
| `garch_t` | The same volatility model with Student-t innovations (quantile rescaled to unit variance) |

**Conventions.** VaR is a positive loss (0.02 means a 2% loss). An *exception* is a day when the return falls below −VaR. The forecast for day *t* never uses day *t* or later, which is the most common bug in backtests, so it is worth re-reading the code below with that in mind.
